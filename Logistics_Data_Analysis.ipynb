#### Outlier Detection

```python
Q1=data[['distance_km','package_weight_kg','delivery_time_hours','expected_time_hours','delivery_rating','delivery_cost']].quantile(0.25)
Q1
```

```python
Q3=data[['distance_km','package_weight_kg','delivery_time_hours','expected_time_hours','delivery_rating','delivery_cost']].quantile(0.75)
Q3
```

```python
IQR=Q3-Q1  #Interquartile Range
IQR
```

```python
lower=Q1-(1.5 *IQR)
lower
```

```python
upper=Q3+ (1.5 *IQR)
upper
```

```python
outlier_counts=((data[Q1.index]<lower) | (data[Q3.index]>upper)).sum()
outlier_counts
```

```python
outlier_delivery_time=data[(data['delivery_time_hours']<lower['delivery_time_hours']) | (data['delivery_time_hours']>upper['delivery_time_hours'])]
outlier_delivery_time
```
