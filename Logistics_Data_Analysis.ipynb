## 9. Justification of Visualization Methods

1. **Delivery Status Bar Chart:** A bar chart was selected because delivery status is a categorical variable. It makes it easy to compare the counts of delivered, delayed, and failed shipments.

2. **Delivery Time Histogram:** A histogram was selected because delivery time is a numerical variable. Grouping values into intervals helps show the frequency distribution and identify common delivery durations.

3. **Distance vs Delivery Time Scatter Plot:** A scatter plot was selected because both distance and delivery time are numerical variables. It helps reveal relationships, trends, and the spread of observations.

4. **Average Delivery Cost by Vehicle Type Bar Chart:** A bar chart was selected because vehicle type is categorical and average delivery cost is numerical. It allows direct comparison of average costs between vehicle categories.

5. **Delay Rate by Weather Condition Bar Chart:** A bar chart was selected to compare calculated delay percentages across weather categories and identify differences in observed delay rates.

6. **Correlation Heatmap:** A heatmap was selected because it displays correlation coefficients between several numerical variables in one figure, making stronger positive and negative relationships easier to identify.

## 10. Analytical Recommendations

Based on the findings from the logistics dataset, the following recommendations are proposed:

1. **Improve Weather-Based Planning:** Investigate delivery delays during stormy and rainy weather. Consider reviewing route planning and delivery schedules during adverse weather conditions.

2. **Review Express Delivery Operations:** Express deliveries showed a time-based delay rate of 64.67%. Investigate the reasons for this high rate and review whether delivery schedules and operational processes need improvement.

3. **Investigate Delivery Failures:** Xpressbees recorded 167 failed deliveries in the dataset. Review these records to identify possible reasons for unsuccessful deliveries and opportunities to improve delivery completion.

4. **Review Transportation Costs:** Distance and delivery cost showed a correlation of 0.990772. Analyze the relationship further to understand transportation cost patterns and possible cost-saving opportunities.

5. **Improve Customer Experience:** Delivery time and delivery rating showed a negative correlation of -0.27. Investigate whether longer delivery times are associated with lower customer ratings.

These recommendations identify areas for further investigation. The observed relationships do not establish that a particular factor directly causes delays, failures, or higher costs.
